## Import Required Libraries

```python
import os
import re

import numpy as np
import pandas as pd

from PyPDF2 import PdfReader

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

print("Libraries imported successfully!")
```

## Take Resume and Job Description

```python
resume_path = input(
    "Enter the path of your resume PDF: "
).strip()

job_description = input(
    "Paste the job description here: "
).strip()


if not os.path.exists(resume_path):
    print("Resume file not found!")

else:
    print("\nResume found successfully!")

print("\nJob description received!")
```

## Extract Text from Resume

```python
reader = PdfReader(resume_path)

resume_text = ""

for page in reader.pages:

    text = page.extract_text()

    if text:
        resume_text += text


print("Resume text extracted successfully!")
```

## Clean and Preprocess Text

```python
def clean_text(text):

    text = text.lower()

    text = re.sub(
        r'[^a-zA-Z0-9\s]',
        ' ',
        text
    )

    text = re.sub(
        r'\s+',
        ' ',
        text
    )

    return text.strip()


clean_resume = clean_text(
    resume_text
)

clean_job_description = clean_text(
    job_description
)


print("Text preprocessing completed!")
```

## Extract Skills

```python
skills_list = [
    "python",
    "java",
    "c++",
    "sql",
    "machine learning",
    "deep learning",
    "artificial intelligence",
    "nlp",
    "tensorflow",
    "pytorch",
    "scikit learn",
    "pandas",
    "numpy",
    "matplotlib",
    "opencv",
    "git",
    "github",
    "docker",
    "fastapi",
    "flask",
    "streamlit",
    "langchain",
    "rag",
    "llm",
    "aws",
    "power bi",
    "excel"
]


resume_skills = []
job_skills = []


for skill in skills_list:

    if skill in clean_resume:
        resume_skills.append(skill)

    if skill in clean_job_description:
        job_skills.append(skill)


print("Skills found in Resume:")
print(resume_skills)

print("\nSkills required in Job Description:")
print(job_skills)
```

## Calculate Resume-JD Similarity

```python
documents = [
    clean_resume,
    clean_job_description
]


vectorizer = TfidfVectorizer(
    stop_words="english"
)


tfidf_matrix = vectorizer.fit_transform(
    documents
)


similarity = cosine_similarity(
    tfidf_matrix[0:1],
    tfidf_matrix[1:2]
)


similarity_score = similarity[0][0] * 100


print(
    f"Resume-JD Similarity: "
    f"{similarity_score:.2f}%"
)
```

## Find Matched and Missing Skills

```python
matched_skills = []

missing_skills = []


for skill in job_skills:

    if skill in resume_skills:
        matched_skills.append(skill)

    else:
        missing_skills.append(skill)


print("Matched Skills:")
print(matched_skills)


print("\nMissing Skills:")
print(missing_skills)
```

## Generate Suggestions

```python
suggestions = []


if len(missing_skills) > 0:

    for skill in missing_skills:

        suggestions.append(
            f"Consider adding or learning {skill} "
            f"if it is relevant to your experience."
        )

else:

    suggestions.append(
        "Your resume covers all detected job skills."
    )


print("Resume Suggestions:\n")

for suggestion in suggestions:

    print("-", suggestion)
```

## Calculate Overall Resume Score

```python
if len(job_skills) > 0:

    skill_score = (
        len(matched_skills)
        / len(job_skills)
    ) * 100

else:

    skill_score = 0


overall_score = (
    (similarity_score * 0.5)
    +
    (skill_score * 0.5)
)


print(
    f"Text Similarity Score: "
    f"{similarity_score:.2f}%"
)

print(
    f"Skill Match Score: "
    f"{skill_score:.2f}%"
)

print(
    f"Overall Resume Score: "
    f"{overall_score:.2f}%"
)
```

## Display Complete Analysis

```python
print("\n")
print("=" * 55)
print("           AI RESUME ANALYZER")
print("=" * 55)


print(
    f"\nOverall Resume Score: "
    f"{overall_score:.2f}%"
)


print(
    f"\nResume-JD Similarity: "
    f"{similarity_score:.2f}%"
)


print(
    f"\nSkill Match Score: "
    f"{skill_score:.2f}%"
)


print("\nMatched Skills:")

if matched_skills:

    for skill in matched_skills:
        print("-", skill)

else:

    print("No matched skills found.")


print("\nMissing Skills:")

if missing_skills:

    for skill in missing_skills:
        print("-", skill)

else:

    print("No missing skills found.")


print("\nSuggestions:")

for suggestion in suggestions:

    print("-", suggestion)


print("\n" + "=" * 55)
print("             ANALYSIS COMPLETE")
print("=" * 55)
```
