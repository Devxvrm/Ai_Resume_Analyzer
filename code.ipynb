## Import Required Libraries

```python
import os
import re

import numpy as np
import pandas as pd

from PyPDF2 import PdfReader

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

print("Libraries imported successfully!")
```

## Take Resume and Job Description

```python
resume_path = input(
    "Enter the path of your resume PDF: "
).strip()

job_description = input(
    "Paste the job description here: "
).strip()


if not os.path.exists(resume_path):
    print("Resume file not found!")

else:
    print("\nResume found successfully!")

print("\nJob description received!")
```

## Extract Text from Resume

```python
reader = PdfReader(resume_path)

resume_text = ""

for page in reader.pages:

    text = page.extract_text()

    if text:
        resume_text += text


print("Resume text extracted successfully!")

print("\nFirst 500 characters:\n")
print(resume_text[:500])
```

## Clean and Preprocess Text

```python
def clean_text(text):

    text = text.lower()

    text = re.sub(
        r'[^a-zA-Z0-9\s]',
        ' ',
        text
    )

    text = re.sub(
        r'\s+',
        ' ',
        text
    )

    return text.strip()


clean_resume = clean_text(
    resume_text
)

clean_job_description = clean_text(
    job_description
)


print("Text preprocessing completed!")

print("\nClean Resume Text:\n")
print(clean_resume[:500])
```

## Extract Skills

```python
skills_list = [
    "python",
    "java",
    "c++",
    "sql",
    "machine learning",
    "deep learning",
    "artificial intelligence",
    "nlp",
    "tensorflow",
    "pytorch",
    "scikit learn",
    "pandas",
    "numpy",
    "matplotlib",
    "opencv",
    "git",
    "github",
    "docker",
    "fastapi",
    "flask",
    "streamlit",
    "langchain",
    "rag",
    "llm",
    "aws",
    "power bi",
    "excel"
]


resume_skills = []
job_skills = []


for skill in skills_list:

    if skill in clean_resume:
        resume_skills.append(skill)

    if skill in clean_job_description:
        job_skills.append(skill)


print("Skills found in Resume:")
print(resume_skills)

print("\nSkills required in Job Description:")
print(job_skills)
```

## Calculate Resume-JD Similarity

```python
documents = [
    clean_resume,
    clean_job_description
]


vectorizer = TfidfVectorizer(
    stop_words="english"
)


tfidf_matrix = vectorizer.fit_transform(
    documents
)


similarity = cosine_similarity(
    tfidf_matrix[0:1],
    tfidf_matrix[1:2]
)


similarity_score = similarity[0][0] * 100


print(
    f"Resume-JD Similarity: "
    f"{similarity_score:.2f}%"
)
```
